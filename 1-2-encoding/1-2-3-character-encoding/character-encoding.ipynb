# Character Encoding

```python
def save_to_computer(text):
    raw = text.encode("utf-8")
    return raw
```

# simulate "garbled text"

```python
text = "你好, 欢迎来到字符编码课堂"
print(text)
```

```python
encoded_text = save_to_computer(text)
print(encoded_text)
```

```python
print(encoded_text.hex(" ").upper())
```

```python
decoded_text = encoded_text.decode("gb18030")
print(decoded_text)
```
