### 操作顺序（约5分钟）
1. 选择 **① 原波形**：指出横轴（时间）、纵轴（幅度）。
2. 选择 **② 采样**：固定3 bit，逐步改变采样率，比较横向样本点数量。
3. 选择 **③ 量化**：固定采样率，比较2、3、4 bit对应的纵向可选等级。
4. 选择 **④ 编码**：点击任一蓝色样本点，观察相应的二进制编码带，反过来也可点击码字选中样本。
5. 查看**量化等级对照表**，解释选中样本属于哪个输入范围、代表值是多少。
6. 最后让学生复述：采样回答“何时记录”，量化回答“记录成哪个近似值”，编码回答“如何表示等级编号”。

**规则**：采用均匀mid-rise量化，仅作教学实例；同一采样值可依据所选位数落入不同区间。灰色平滑线是预先给定的信号参考，并非用采样点简单连线重建的波形。

```python
# 可选：在Mac内核直接检查一个量化实例，显示模型的数据依据。
from lesson_audio_v7 import quantize
for bits in (2, 3, 4):
    level, value = quantize(0.37, bits)
    print(f"{bits} bit：等级 {level}，代表值 {value:.3f}，码字 {level:0{bits}b}")
```
